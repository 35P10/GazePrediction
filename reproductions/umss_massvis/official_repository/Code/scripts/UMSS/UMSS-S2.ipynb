## Sampling Process

```python
def predict(imgpred):
    """
        Predict 40 scanpaths given an image
        Params:
            image
    """

    array_shape = (480, 640)

    def sample_slice(pred, size, n_samples=1, use_heuristic=True, samples=[]):
        """
            Takes random points from the slice
            taking into account the probabilities of each
            pixel    
            Params:
                pred = 2d array with probability of each pixel      
                size = (height, width) of the image
                n_samples = number of samples taken from this slice
        """ 
        
        def num_to_pos(n, size):
            """
                Convert an index of the flatten 2d image to a 
                coordenate (x, y)
                Params: 
                    size = (height, width)
            """
            # x and y pos
            x = n % array_shape[1]
            y = n / array_shape[1]
            return (x, y)

        pred[pred < np.max(pred) / 2] = 0

        # Normalize to predictions
        p = pred / np.sum(pred)

        

        # Flatten
        p = p.flatten()

        # Store the samples 
        #samples = []
        for i in range(int(n_samples)):
            if samples and use_heuristic:
                #print('samples shape:', np.array(samples).shape, samples[-1])

                # Original Saltinet sampling
                gaussian = np.zeros(array_shape)
                gaussian[int(samples[-1][1]), int(samples[-1][0])] = 1
                mask = ndimage.filters.gaussian_filter(gaussian, [200, 200])

                #print(boxes)
                
                
                p = pred / np.sum(pred)
                p =  p * mask
                p = p / np.sum(p)
                p = p.flatten()         

            n = np.random.choice(array_shape[0] * array_shape[1], p=p)
            pos = num_to_pos(n, size=size)
            samples.append(pos) 

        return samples

    def sample_volume(vol, n_samples=24, size=array_shape, use_heuristic=True):
        # Choose how many samples take per slice
        # i.e. 
        #     n_slices = 8 , n_samples = 12  => samples_per_slice = [2, 2, 2, 2, 1, 1, 1, 1]   
        n_slices = vol.shape[0]

        samples_per_slice = []
        dur = np.zeros(n_samples)
        accum = 0
        last_index = 0
        b1 = False
        b2 = False
        for i in range(n_samples):
            t = get_duration_fixation()
            dur[i] = t
            # 0.5s
            if np.sum(dur) > 0.5 and not b1:
                b1 = True
                samples_per_slice.append(i)
                last_index = i
            # 3s
            elif np.sum(dur) > 3 and not b2:
                b2 = True
                samples_per_slice.append(i - last_index)
                last_index = i

        samples_per_slice.append(n_samples - last_index)
                
        
        # Sample each slice of the volume
        samples = []
        for i, n_samples in enumerate(samples_per_slice):
            samples = sample_slice(vol[i], size, n_samples=n_samples, use_heuristic=use_heuristic, samples=samples)
  
        # Normalize positions
        for i in range(len(samples)):
            x = int((float(samples[i][0]) / array_shape[1]) * size[1])
            y = int((float(samples[i][1]) / array_shape[0]) * size[0])
            samples[i] = (x, y, dur[i]) 
        return samples

    #print('imgpred shape:', np.array(imgpred).shape)

    scanpaths = []
    for i in range(40):
        n_samples = get_number_fixations()
        # print(i,n_samples)
        # s: (x, y, t)
        s = sample_volume(imgpred, n_samples=n_samples , size=array_shape, use_heuristic=True)

        for j in range(n_samples):
            # [user, index, time, x, y]
            pos_1s = [i+1, j+1, s[j][2], s[j][0], s[j][1]]
            scanpaths.append(pos_1s)

    # Generate a np.array to output
    scanpaths_array = np.array(scanpaths)

    return scanpaths_array
```

```python
def predict_and_save():
    """ 
        Predicts multiple images and saves them in .csv format
    on an output path
    """

    # Preproces and load images

    for i in range(0, len(salmap_paths), 1):
        name = salmap_paths[i].replace(salmap_path+'/','').replace('_500', '')
        print('Working on image %s, %d of %d' % (name, i+1, len(salmap_paths)))

        # Predict the scanpaths
        scanpaths = predict(preds[:,i,:,:])

        # Turn into a float np.array
        # scanpaths = np.array(scanpaths) # dtype=np.float32

        # Save in output folder
        np.savetxt('%s/%s.csv'%(output_path, name), scanpaths, fmt=('%i','%i','%.06f','%i','%i'), header="user, index, time, x, y", delimiter=',')

        print('Saved scanpaths in file %s/%s.csv' %(output_path, name))

    print('Done!')
    return True
```
